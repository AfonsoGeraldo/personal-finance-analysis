```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../data/transactions.csv')
print(df.head())
```

```python
df.info()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df.info()
```

```python
df = df.set_index('Date')
df.head()
```

```python
monthly_totals = df['Amount'].resample('ME').sum()
monthly_totals.head(12)
```

```python
df_income = df[df['Type'] == 'Income']
df_expenses = df[df['Type'] == 'Expense']
df_income.head()
df_expenses.head()
```

```python
monthly_income = df_income['Amount'].resample('ME').sum()
monthly_expenses = df_expenses['Amount'].resample('ME').sum()
print('Monthly Income:')
print(monthly_income.head(12))
print('Monthly Expenses:')
print(monthly_expenses.head(12))
```

```python
monthly_net = monthly_income - monthly_expenses
print('Monthly Net:') 
print(monthly_net.head(12))
```

```python
category_totals = df_expenses.groupby('Category')['Amount'].sum().sort_values(ascending=False)
print('Category Totals:')
print(category_totals)
```
