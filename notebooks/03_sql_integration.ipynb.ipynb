```python
from sqlalchemy import create_engine

engine = create_engine(
    'mssql+pyodbc://localhost/Testes?driver=ODBC+Driver+18+for+SQL+Server&trusted_connection=yes&TrustServerCertificate=yes'
)
```

First - Read FROM SQL

```python
import pandas as pd

query = "SELECT * FROM Transactions"
df_sql = pd.read_sql(query, engine)

print(df_sql)
```

Write data to SQL Server

```python
df_summary = pd.read_csv('../data/week2.csv')
print(df_summary)

df_summary.to_sql('WeekendExpenseSummary', engine, if_exists='replace', index=False)
print('Done!')
```

Insert with SQLALchemy

```python
from sqlalchemy import text

with engine.connect() as conn:
    conn.execute(text("INSERT INTO WeekendExpenseSummary (Type, Category, [sum], mean) VALUES ('Expense', 'Entertainment', 999.99, 999.99)"))
    conn.commit()
```

Error Handling

```python
from sqlalchemy.exc import SQLAlchemyError

try:
    pd.read_sql('SELECT * FROM ABC', engine)
except SQLAlchemyError as e:
    print("Something happened: ", e)
except Exception as e:
    print("Something happened: ", e)
```
