```python
import pandas as pd

df = pd.read_csv('../data/transactions.csv')
df.head()
```

loc/iloc - Exercise 1

```python
#Using .loc, select the 'Category' and 'Amount' columns, but only for rows where Amount > 200.

loc1 = df.loc[df['Amount'] > 200,['Amount', 'Category']]

print(loc1)
```

# 2. Using .iloc, select just the first 5 rows and the first 3 columns of df (whatever those columns are positionally — don't name them, use position).

```python
iloc2 = df.iloc[0:5, 0:3]
print(iloc2)
```

# 3. Using .loc, select the 'Description' for all rows where Category == 'Housing'.

```python
loc3 = df.loc[df['Category'] == 'Housing', 'Description']
print(loc3)
```

# 4. Combine what you already know: using .loc with a boolean condition on one column, but returning a different column — select the 'Date' of every row where Type == 'Income'.

```python
loc4 = df.loc[df['Type'] == 'Income', 'Date']
print(loc4)
```

```python
print(df.iloc[0:5])
print(df.loc[0:5])
```

# 6. Using .agg(), for each Category, compute the sum, mean, and count of Amount, all in one call.

```python
groupVar = df.groupby('Category')['Amount'].agg(['sum', 'mean', 'count'])
print(groupVar)
```

```python
groupVar = df.groupby(['Category', 'Type'])['Amount'].sum()
print(groupVar)
```
