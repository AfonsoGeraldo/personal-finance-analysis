1. Read and build a df

```python
import pandas as pd

df = pd.read_csv('../data/transactions.csv')
df.head()
```

2. Clean - Remove Bonus

```python
df = df[df['Category'] != 'Bonus']

# Just to check - it is empty
print(df[df['Category'] == 'Bonus'])
```

3. Aggregate

```python
print(f"Before: {df['Date'].dtype}")

df['Date'] = pd.to_datetime(df['Date'])

print(f"After: {df['Date'].dtype}")

df['IsWeekend'] = (df['Date'].dt.day_of_week == 5) | (df['Date'].dt.day_of_week == 6)

sum_amount = df[df['IsWeekend']].groupby(['Type', 'Category'])['Amount'].agg(['sum', 'mean'])

print(sum_amount)
```

4. Export

```python
sum_amount.to_csv('../data/week2.csv', index=True)
print("Week 2 finished!")
```
