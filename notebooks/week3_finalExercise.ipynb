```python
import pandas as pd
```

```python
from sqlalchemy import create_engine

engine = create_engine(
    'mssql+pyodbc://localhost/Testes?driver=ODBC+Driver+18+for+SQL+Server&trusted_connection=yes&TrustServerCertificate=yes'
)
```

```python
try:
    query = 'Select * from dbo.loans'
    df = pd.read_sql(query, engine)
    df.head()

    #clean
    df = df[df['OutstandingBalance'] != 0]

    #groupBy
    df_groupped = df.groupby('RiskCategory')[['OutstandingBalance']].mean().reset_index()

    #save
    df_groupped.to_sql('AverageOutstandingBalance', engine, if_exists='replace', index=False)

except Exception as e:

    print(f"There's an error: {e}")
```
